```python
import os
import time
import gc
import random
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
)
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from xgboost import XGBClassifier
from imblearn.over_sampling import SMOTE

import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout, Conv1D, MaxPooling1D, Flatten
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# SETTINGS — CHANGE THESE FIRST
# ============================================================

DATA_FILE = "RT_IOT2022.csv"
TARGET = "Attack_type"

RANDOM_STATE = 42
TEST_SIZE = 0.20
N_FEATURES = 20

# Fast mode:
RUN_WITHOUT_SMOTE = True
RUN_WITH_SMOTE = True

# SMOTE is applied to ALL remaining classical models.
# SVM has been removed.
SMOTE_MODELS = {"XGBoost", "Decision Tree", "Random Forest", "KNN"}

# Confusion matrices are generated for EVERY experiment.
PLOT_CONFUSION_MATRICES = True

# Set True for the final run if you want all comparison graphs.
PLOT_FINAL_GRAPHS = True

# Classical ML speed settings
XGB_TREES = 50
XGB_DEPTH = 4
RF_TREES = 50
DT_MAX_DEPTH = 15

# Neural network speed settings
NN_EPOCHS = 15
NN_BATCH_SIZE = 128
NN_PATIENCE = 3

# ============================================================
# REPRODUCIBILITY / GPU
# ============================================================

np.random.seed(RANDOM_STATE)
random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

gpus = tf.config.list_physical_devices("GPU")
print("TensorFlow GPUs:", gpus if gpus else "None (CPU mode)")

if gpus:
    for gpu in gpus:
        try:
            tf.config.experimental.set_memory_growth(gpu, True)
        except Exception:
            pass

# ============================================================
# LOAD DATA
# ============================================================

start_total = time.time()

df = pd.read_csv(DATA_FILE)

df = df.replace([np.inf, -np.inf], np.nan)
df = df.fillna(0)

print("=" * 80)
print("DATASET INFORMATION")
print("=" * 80)
print("Dataset shape:", df.shape)
print("Target:", TARGET)

# ============================================================
# TARGET
# ============================================================

label_encoder = LabelEncoder()
y = label_encoder.fit_transform(df[TARGET].astype(str))

print("\nClasses:")
for i, name in enumerate(label_encoder.classes_):
    print(i, "=", name)

print("\nOriginal class distribution:")
print(df[TARGET].value_counts())

# ============================================================
# OPTIONAL CLASS DISTRIBUTION GRAPH
# ============================================================

plt.figure(figsize=(12, 6))
df[TARGET].value_counts().plot(kind="bar")
plt.title("RT-IoT2022 Attack Type Distribution")
plt.xlabel("Attack Type")
plt.ylabel("Number of Samples")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

# ============================================================
# FEATURES
# ============================================================

X = df.drop(
    columns=[TARGET, "Unnamed: 0"],
    errors="ignore"
).copy()

categorical_columns = X.select_dtypes(include=["object"]).columns

print("\nCategorical features:")
print(categorical_columns.tolist())

# Frequency encoding
for col in categorical_columns:
    frequencies = X[col].astype(str).value_counts()
    X[col] = X[col].astype(str).map(frequencies)

X = X.apply(pd.to_numeric, errors="coerce")
X = X.replace([np.inf, -np.inf], np.nan).fillna(0)

print("\nFinal feature shape:", X.shape)

# ============================================================
# TRAIN / TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=RANDOM_STATE,
    stratify=y,
)

print("Training data:", X_train.shape)
print("Testing data :", X_test.shape)

# ============================================================
# FEATURE SELECTION
# ============================================================

k = min(N_FEATURES, X_train.shape[1])

selector = SelectKBest(
    score_func=f_classif,
    k=k
)

X_train_selected = selector.fit_transform(X_train, y_train)
X_test_selected = selector.transform(X_test)

selected_features = X.columns[selector.get_support()]

print("\nSelected features:")
for feature in selected_features:
    print(" -", feature)

# ============================================================
# SCALE DATA
# ============================================================

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_selected).astype(np.float32)
X_test_scaled = scaler.transform(X_test_selected).astype(np.float32)

# Free memory from intermediate objects
del X_train_selected, X_test_selected
gc.collect()

# ============================================================
# SMOTE
# ============================================================

print("\nTraining class distribution BEFORE SMOTE:")
unique_before, counts_before = np.unique(y_train, return_counts=True)

for cls, count in zip(unique_before, counts_before):
    print(label_encoder.inverse_transform([cls])[0], ":", count)

X_train_smote = None
y_train_smote = None

if RUN_WITH_SMOTE:
    minimum_class_count = counts_before.min()
    smote_k = min(5, minimum_class_count - 1)

    if smote_k >= 1:
        print("\nApplying SMOTE...")
        smote = SMOTE(
            random_state=RANDOM_STATE,
            k_neighbors=smote_k,
        )

        smote_start = time.time()
        X_train_smote, y_train_smote = smote.fit_resample(
            X_train_scaled,
            y_train
        )

        print(
            "SMOTE time:",
            round(time.time() - smote_start, 2),
            "seconds"
        )

        print(
            "SMOTE training shape:",
            X_train_smote.shape
        )

    else:
        print(
            "SMOTE skipped: a class has fewer than 2 training samples."
        )

# ============================================================
# RESULTS
# ============================================================

results = []


def evaluate_model(
    model_name,
    experiment,
    y_true,
    y_pred,
    elapsed_time,
):
    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        average="weighted",
        zero_division=0
    )

    print("\n" + "=" * 80)
    print(model_name, "-", experiment)
    print("=" * 80)

    print(f"Accuracy : {accuracy:.4f}")
    print(f"Precision: {precision:.4f}")
    print(f"Recall   : {recall:.4f}")
    print(f"F1 Score : {f1:.4f}")
    print(f"Time (s) : {elapsed_time:.2f}")

    # ========================================================
    # CLASSIFICATION REPORT
    # ========================================================

    print("\nClassification Report:")

    print(
        classification_report(
            y_true,
            y_pred,
            target_names=label_encoder.classes_,
            zero_division=0,
        )
    )

    # ========================================================
    # CONFUSION MATRIX
    # ========================================================

    print("\nConfusion Matrix:")

    cm = confusion_matrix(
        y_true,
        y_pred
    )

    print(cm)

    results.append([
        model_name,
        experiment,
        accuracy,
        precision,
        recall,
        f1,
        elapsed_time,
    ])

    # ========================================================
    # CONFUSION MATRIX PLOT
    # ========================================================

    if PLOT_CONFUSION_MATRICES:

        fig, ax = plt.subplots(
            figsize=(10, 8)
        )

        disp = ConfusionMatrixDisplay(
            confusion_matrix=cm,
            display_labels=label_encoder.classes_,
        )

        disp.plot(
            ax=ax,
            xticks_rotation=45
        )

        plt.title(
            f"{model_name} - {experiment}"
        )

        plt.xlabel("Predicted Label")
        plt.ylabel("True Label")

        plt.tight_layout()
        plt.show()


def run_ml_model(
    model,
    model_name,
    X_train_data,
    y_train_data,
    experiment,
):
    print(
        f"\nTraining: {model_name} | {experiment}"
    )

    start = time.time()

    model.fit(
        X_train_data,
        y_train_data
    )

    y_pred = model.predict(
        X_test_scaled
    )

    elapsed = time.time() - start

    evaluate_model(
        model_name,
        experiment,
        y_test,
        y_pred,
        elapsed,
    )

    return model


# ============================================================
# MODEL FACTORIES
# ============================================================

def make_models():

    return {

        "XGBoost": XGBClassifier(
            n_estimators=XGB_TREES,
            learning_rate=0.1,
            max_depth=XGB_DEPTH,
            random_state=RANDOM_STATE,
            n_jobs=-1,
            eval_metric="mlogloss",
        ),

        "Decision Tree": DecisionTreeClassifier(
            random_state=RANDOM_STATE,
            max_depth=DT_MAX_DEPTH,
        ),

        "Random Forest": RandomForestClassifier(
            n_estimators=RF_TREES,
            random_state=RANDOM_STATE,
            n_jobs=-1,
        ),

        "KNN": KNeighborsClassifier(
            n_neighbors=5,
            n_jobs=-1,
        ),
    }


# ============================================================
# CLASSICAL ML
# ============================================================

print("\n" + "#" * 80)
print("FAST CLASSICAL MACHINE LEARNING")
print("#" * 80)

# ============================================================
# WITHOUT SMOTE
# ============================================================

if RUN_WITHOUT_SMOTE:

    print("\n--- WITHOUT SMOTE ---")

    for name, model in make_models().items():

        run_ml_model(
            model,
            name,
            X_train_scaled,
            y_train,
            "Without SMOTE",
        )


# ============================================================
# WITH SMOTE
# ============================================================

if RUN_WITH_SMOTE and X_train_smote is not None:

    print("\n--- WITH SMOTE ---")

    # ALL CLASSICAL MODELS RUN
    # NO MODEL IS SKIPPED

    for name, model in make_models().items():

        run_ml_model(
            model,
            name,
            X_train_smote,
            y_train_smote,
            "With SMOTE",
        )


# ============================================================
# RANDOM FOREST FEATURE IMPORTANCE
# ============================================================

# Train one RF on SMOTE data if available; otherwise use normal data.

rf_train_X = (
    X_train_smote
    if X_train_smote is not None
    else X_train_scaled
)

rf_train_y = (
    y_train_smote
    if y_train_smote is not None
    else y_train
)

rf_importance = RandomForestClassifier(
    n_estimators=RF_TREES,
    random_state=RANDOM_STATE,
    n_jobs=-1,
)

rf_importance.fit(
    rf_train_X,
    rf_train_y
)

feature_importance_df = pd.DataFrame({
    "Feature": selected_features,
    "Importance": rf_importance.feature_importances_,
})

top10 = (
    feature_importance_df
    .sort_values(
        by="Importance",
        ascending=False
    )
    .head(10)
)

print("\n" + "=" * 80)
print("TOP 10 RANDOM FOREST FEATURES")
print("=" * 80)

print(top10)

plt.figure(
    figsize=(10, 6)
)

plt.barh(
    top10["Feature"][::-1],
    top10["Importance"][::-1]
)

plt.xlabel(
    "Feature Importance"
)

plt.ylabel(
    "Feature"
)

plt.title(
    "Top 10 Features - Random Forest"
)

plt.tight_layout()
plt.show()


# ============================================================
# DEEP LEARNING
# ============================================================

num_features = X_train_scaled.shape[1]
num_classes = len(np.unique(y))


def create_mlp(
    input_features,
    classes
):

    model = Sequential([
        Input(
            shape=(input_features,)
        ),

        Dense(
            128,
            activation="relu"
        ),

        Dropout(
            0.3
        ),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(
            0.2
        ),

        Dense(
            32,
            activation="relu"
        ),

        Dense(
            classes,
            activation="softmax"
        ),
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )

    return model


def create_cnn(
    input_features,
    classes
):

    model = Sequential([
        Input(
            shape=(input_features, 1)
        ),

        Conv1D(
            filters=64,
            kernel_size=3,
            activation="relu",
            padding="same",
        ),

        MaxPooling1D(
            pool_size=2
        ),

        Conv1D(
            filters=32,
            kernel_size=3,
            activation="relu",
            padding="same",
        ),

        MaxPooling1D(
            pool_size=2
        ),

        Flatten(),

        Dense(
            64,
            activation="relu"
        ),

        Dropout(
            0.3
        ),

        Dense(
            classes,
            activation="softmax"
        ),
    ])

    model.compile(
        optimizer="adam",
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"],
    )

    return model


def run_deep_learning(
    model,
    model_name,
    X_train_data,
    y_train_data,
    X_test_data,
    experiment,
    cnn=False,
):

    print(
        f"\nTraining: {model_name} | {experiment}"
    )

    start = time.time()

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=NN_PATIENCE,
        restore_best_weights=True,
    )

    if cnn:

        X_train_input = (
            X_train_data[..., np.newaxis]
        )

        X_test_input = (
            X_test_data[..., np.newaxis]
        )

    else:

        X_train_input = X_train_data
        X_test_input = X_test_data

    model.fit(
        X_train_input,
        y_train_data,
        validation_split=0.1,
        epochs=NN_EPOCHS,
        batch_size=NN_BATCH_SIZE,
        callbacks=[early_stop],
        verbose=0,
    )

    probabilities = model.predict(
        X_test_input,
        verbose=0,
        batch_size=NN_BATCH_SIZE,
    )

    y_pred = np.argmax(
        probabilities,
        axis=1
    )

    elapsed = time.time() - start

    evaluate_model(
        model_name,
        experiment,
        y_test,
        y_pred,
        elapsed,
    )

    # Release TensorFlow graph memory between models.

    tf.keras.backend.clear_session()

    gc.collect()


# ============================================================
# RUN MLP + CNN
# ============================================================

print("\n" + "#" * 80)
print("FAST DEEP LEARNING")
print("#" * 80)


# ============================================================
# MLP + CNN WITHOUT SMOTE
# ============================================================

if RUN_WITHOUT_SMOTE:

    # ----------------------------
    # MLP WITHOUT SMOTE
    # ----------------------------

    mlp = create_mlp(
        num_features,
        num_classes
    )

    run_deep_learning(
        mlp,
        "MLP",
        X_train_scaled,
        y_train,
        X_test_scaled,
        "Without SMOTE",
        cnn=False,
    )

    # ----------------------------
    # CNN WITHOUT SMOTE
    # ----------------------------

    cnn = create_cnn(
        num_features,
        num_classes
    )

    run_deep_learning(
        cnn,
        "1D CNN",
        X_train_scaled,
        y_train,
        X_test_scaled,
        "Without SMOTE",
        cnn=True,
    )


# ============================================================
# MLP + CNN WITH SMOTE
# ============================================================

if RUN_WITH_SMOTE and X_train_smote is not None:

    # SMOTE neural networks are enabled.
    # No model is skipped.

    RUN_SMOTE_NN = True

    if RUN_SMOTE_NN:

        # ----------------------------
        # MLP WITH SMOTE
        # ----------------------------

        mlp_smote = create_mlp(
            num_features,
            num_classes
        )

        run_deep_learning(
            mlp_smote,
            "MLP",
            X_train_smote,
            y_train_smote,
            X_test_scaled,
            "With SMOTE",
            cnn=False,
        )

        # ----------------------------
        # CNN WITH SMOTE
        # ----------------------------

        cnn_smote = create_cnn(
            num_features,
            num_classes
        )

        run_deep_learning(
            cnn_smote,
            "1D CNN",
            X_train_smote,
            y_train_smote,
            X_test_scaled,
            "With SMOTE",
            cnn=True,
        )


# ============================================================
# FINAL RESULTS
# ============================================================

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "SMOTE",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "Time (s)",
    ],
)

print(
    "\n" + "=" * 100
)

print(
    "FINAL MODEL COMPARISON"
)

print(
    "=" * 100
)

if len(results_df):

    print(
        results_df.to_string(
            index=False
        )
    )

    ranked_results = (
        results_df
        .sort_values(
            by="F1 Score",
            ascending=False,
        )
    )

    print(
        "\n" + "=" * 100
    )

    print(
        "MODELS RANKED BY F1 SCORE"
    )

    print(
        "=" * 100
    )

    print(
        ranked_results.to_string(
            index=False
        )
    )

    results_df.to_csv(
        "RT_IOT2022_classification_results_fast.csv",
        index=False,
    )


# ============================================================
# FINAL GRAPHS
# ============================================================

if PLOT_FINAL_GRAPHS and len(results_df):

    labels = (
        results_df["Model"]
        + "\n"
        + results_df["SMOTE"]
    )

    metrics = [
        (
            "Accuracy",
            "Model Accuracy Comparison"
        ),

        (
            "F1 Score",
            "Model F1 Score Comparison"
        ),

        (
            "Precision",
            "Model Precision Comparison"
        ),

        (
            "Recall",
            "Model Recall Comparison"
        ),

        (
            "Time (s)",
            "Training Time Comparison"
        ),
    ]

    for metric, title in metrics:

        plt.figure(
            figsize=(14, 7)
        )

        plt.bar(
            labels,
            results_df[metric]
        )

        plt.ylabel(
            metric
        )

        plt.title(
            title
        )

        plt.xticks(
            rotation=45,
            ha="right"
        )

        plt.tight_layout()

        plt.show()


print(
    "\nTotal runtime:",
    round(
        time.time() - start_total,
        2
    ),
    "seconds"
)

print(
    "\nDone."
)
```
